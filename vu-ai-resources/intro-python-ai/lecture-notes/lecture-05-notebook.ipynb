```python
my_lst = [4,5,6]
def exists_in(lst: list, number: int) -> bool:
    for i in lst:
        if i == number:
            return True
    return False 

number_to_check = 4
result = exists_in(my_list, number_to_check)
print(result)
```

you have a new list [30, 20, 15, 9, 50]
make a function that filters: how many people are allowed to drink alcohol: >18 

```python
from typing import List  # Import the List type from the typing module

numbers = [30, 20, 15, 9, 50]  # Renamed my_list to numbers

def filter_numbers(num_list: List[int], threshold: int) -> List[int]:  # Use List with capital L for type hint
    filtered_result = []  # Renamed result to filtered_result
    
    for num in num_list:  # Renamed element to num
        if num >= threshold:  # Changed filter_number to threshold
            filtered_result.append(num)  # Append to filtered_result
    
    return filtered_result  # Indented correctly

# Example usage with the correct function name
print(filter_numbers(numbers, 18))  # Should return [30, 20, 50]
```
