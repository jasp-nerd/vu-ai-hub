Your task is to write a program which uses a function `group_count()` to calculate the number of different ways in which a group (of a given size between 2 and 30) could be constructed from the total students parttaking in the course (i.e., the number of combinations of randomly chosen students). This number is returned in a string containing a message as shown in the example below.

*Hint:* you might have noticed that we included a function `_factorial()` below in the code cell. This will come in handy now: if you want to calculate the factorial of a number *n* for this exercise, you can simply use `_factorial(n)` in your code. Don't worry if the function doesn't make total sense to you right now, as it includes some functionalities and concepts you haven't learnt yet.

Example:

*Input:*
```python
group_size: int = 2
number_of_students: int = 60
```

*Output:*

<pre>
As there are 60 students in the course, groups of 2 students could be constructed in 1770 different ways.
</pre>

````python
group_size: int = 2
number_of_students: int = 60

#// BEGIN_TODO [group_work] Group Work
def _factorial(n):
    factorial = 1
    for num in range(2, n + 1):
        factorial *= num
    return factorial


#// END_TODO [group_work]

group_count(group_size, number_of_students)
````
