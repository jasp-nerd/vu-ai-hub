```python
def find_magic_numbers(numbers: List[int]) -> List[int]:
    """
    Finds all magic numbers (perfect squares) from a list of numbers.
    :Param numbers: A list of integers.
    :Returns: A list containing only the perfect squares.
    """
    magic_numbers = []
    for num in numbers:
        if int(num ** 0.5) ** 2 == num: 
            magic_numbers.append(num)
    return magic_numbers
```
