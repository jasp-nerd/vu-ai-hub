It has two attributes:  
* `position`: a tuple representing the current position of the piece on the board in x, y coordinates (e.g., `(4, 0)` for "e1").  
* `color`: a string representing the color of the piece ("White" or "Black").

### Subclasses:
Each subclass represents a specific type of chess piece. They should initialize just like the base class.

For the movement rules you do not have to worry about the edge of the board or other pieces on the board.

`King`:  
* The `can_move(target_position)` method returns whether the King can move to the `target_position` (one square in any direction).  

`Pawn`:  
* The `can_move(target_position)` method returns whether the Pawn can move to the `target_position`:
  - For White, one square forward (y + 1) or two squares forward from the starting position (y == 1).  
  - For Black, one square forward (y - 1) or two squares forward from the starting position (y == 6).  

### Example
