The following code was taken from [Matplotlib's official documentation](http://matplotlib.org/users/pyplot_tutorial.html).
It uses NumPy and Matplotlib, and it should show a histogram for ten thousand random samples of a normal distribution with a mean of 100 and a standard deviation of 15.

````python
mu, sigma = 100, 15
x = mu + sigma * np.random.randn(10000)

# the histogram of the data
n, bins, patches = plt.hist(x, 50, density=1, facecolor='g', alpha=0.75)

plt.xlabel('Smarts')
plt.ylabel('Probability')
plt.title('Histogram of IQ')
plt.text(60, .025, r'$\mu=100,\ \sigma=15$')
plt.axis([40, 160, 0, 0.03])
plt.grid(True)
plt.show()
````

The following code was taken from the [Pandas' official documentation](http://pandas.pydata.org/pandas-docs/version/0.18.1/visualization.html).
It uses NumPy, Pandas, and Matplotlib, and it should show a time series that accumulates one thousand samples of standard normal distribution, plotting one cumulative value per day, starting at 01 Jan 2000.

````python
ts = pd.Series(np.random.randn(1000), index=pd.date_range('1/1/2000', periods=1000))
ts = ts.cumsum()
ts.plot()
plt.show()
````
